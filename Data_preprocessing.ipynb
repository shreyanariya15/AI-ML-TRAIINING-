```python
df_1 = pd.DataFrame({'ID':[1,2,3,5],
                     'Class':[9,10,2,11]})
df_1
```

```python
df_2 = pd.DataFrame({'ID':[1,2,3,4],
                     'Name':['A','B','C','D']})
df_2
```

```python
merge_df = pd.merge(df_1,df_2,how='left',on='ID',indicator=True)
merge_df
```

```python
merge_df = pd.merge(df_1,df_2,how='inner',on='ID',indicator=True)
merge_df
```

```python
merge_df = pd.merge(df_1,df_2,how='right',on='ID',indicator=True)
merge_df
```

```python
merge_df = pd.merge(df_1,df_2,how='outer',on='ID',indicator=True)
merge_df
```

```python
pd.merge(df_1,df_2,left_index=True,right_index=True)
```

**Fillna() function:**

The fillna() function in pandas is used to fill missing values (NaN) in a DataFrame or Series. You can fill missing values with a specific value, or with a calculated value like the mean, median, or mode of the column. This helps to clean up your data for further analysis.

For example, in the code you have:

df_zero = dataf.fillna(0): This fills all NaN values in the dataf DataFrame with 0.
dataf['Age']=dataf['Age'].fillna(dataf['Age'].mean()): This fills the missing NaN values in the 'Age' column with the mean (average) of the existing 'Age' values.
dataf['City']=dataf['City'].fillna(dataf['City'].mode()[0]): This fills the missing NaN values in the 'City' column with the mode (most frequent value) of the existing 'City' values.
