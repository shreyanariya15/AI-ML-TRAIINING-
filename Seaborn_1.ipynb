```python
import matplotlib.pyplot as plt

death_percentage_by_class = titanic_data.groupby('class')['survived'].apply(lambda x: (1 - x).mean() * 100).reset_index()
death_percentage_by_class.rename(columns={'survived': 'death_percentage'}, inplace=True)


plt.figure(figsize=(8, 6))
sns.barplot(x='class', y='death_percentage', data=death_percentage_by_class, palette='viridis')
plt.title('Percentage of Deaths by Class')
plt.xlabel('Class')
plt.ylabel('Death Percentage (%)')
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for index, row in death_percentage_by_class.iterrows():
    plt.text(index, row['death_percentage'] + 1, f"{row['death_percentage']:.2f}%", color='black',)

plt.show()
```

```python
cars_data = sns.load_dataset('car_crashes')
cars_data
```

```python
crashes = cars_data['total']

sns.lineplot(x='speeding',y='alcohol',data=cars_data)
```

```python
# Plot total crashes vs not_distracted
plt.figure(figsize=(10, 6))
sns.lineplot(x='not_distracted', y='total', data=cars_data)
plt.title('Total Crashes vs Not Distracted Drivers')
plt.xlabel('Not Distracted Drivers (% of drivers involved in fatal collisions)')
plt.ylabel('Total Crashes (fatal collisions per billion vehicle miles)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(x='no_previous', y='total', data=cars_data)
plt.title('Total Crashes vs Drivers with No Previous Accidents')
plt.xlabel('Drivers with No Previous Accidents (% of drivers involved in fatal collisions)')
plt.ylabel('Total Crashes (fatal collisions per billion vehicle miles)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Plot total crashes vs insurance premiums
plt.figure(figsize=(10, 6))
sns.lineplot(x='ins_premium', y='total', data=cars_data)
plt.title('Total Crashes vs Vehicle Insurance Premiums')
plt.xlabel('Vehicle Insurance Premiums (dollars)')
plt.ylabel('Total Crashes (fatal collisions per billion vehicle miles)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
penguins_data = sns.load_dataset('penguins')
penguins_data
```

```python
sns.catplot(data=penguins_data, x='species', y='bill_length_mm', hue='sex', kind='box')
plt.title('Bill Length Distribution by Species and Sex')
plt.xlabel('Species')
plt.ylabel('Bill Length (mm)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=penguins_data, x='bill_length_mm', y='flipper_length_mm', hue='species', style='sex', s=100, alpha=0.8)
plt.title('Bill Length vs. Flipper Length in Penguins')
plt.xlabel('Bill Length (mm)')
plt.ylabel('Flipper Length (mm)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Species/Sex')
plt.show()
```

```python
sns.pairplot(penguins_data, hue='species')
plt.suptitle('Pair Plot of Penguin Data by Species', y=1.02)
plt.show()
```

```python
sns.lineplot(x='sex', y='bill_length_mm', data=penguins_data)
plt.title('Bill Length Distribution by Sex')
plt.xlabel('Sex')
plt.ylabel('Bill Length ')
```
