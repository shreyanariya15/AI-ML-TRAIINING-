```python
heatmap_data = global_warminig.drop(['Indicator Code','Country Code','Indicator Name'],axis=1).set_index('Country Name')
sns.heatmap(heatmap_data)
plt.show()
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(heatmap_data,annot=True)
plt.show()
```

```python
plt.figure(figsize=(16,9))
annot_kws = {'fontsize':10,
             'fontweight':'bold',
             'color': 'black',
             'rotation':'vertical',
             'alpha':0.8,
             'verticalalignment':'center',
             'horizontalalignment':'center',
             'backgroundcolor':'white'

             }
sns.heatmap(heatmap_data,annot=True,annot_kws = annot_kws)
plt.show()
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(heatmap_data.corr())
plt.show()
```
