```python
print('--- Vertical Stacking (np.vstack) ---')

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print('Array 1:', arr1)
print('Array 2:', arr2)


v_stacked_arr = np.vstack((arr1, arr2))
print('\nVertically stacked array (from 1D arrays):\n', v_stacked_arr)

arr3 = np.array([[10, 20, 30]])
arr4 = np.array([[40, 50, 60]])

print('\nArray 3:', arr3)
print('Array 4:', arr4)


v_stacked_arr_2d = np.vstack((arr3, arr4))
print('\nVertically stacked array (from 2D arrays):\n', v_stacked_arr_2d)
```

```python
print('--- Horizontal Stacking (np.hstack) ---')

arr5 = np.array([1, 2, 3])
arr6 = np.array([4, 5, 6])

print('Array 5:', arr5)
print('Array 6:', arr6)

h_stacked_arr = np.hstack((arr5, arr6))
print('\nHorizontally stacked array (from 1D arrays):\n', h_stacked_arr)
arr7 = np.array([[10, 20], [30, 40]])
arr8 = np.array([[50, 60], [70, 80]])

print('\nArray 7:\n', arr7)
print('\nArray 8:\n', arr8)

h_stacked_arr_2d = np.hstack((arr7, arr8))
print('\nHorizontally stacked array (from 2D arrays):\n', h_stacked_arr_2d)
```

## More NumPy Mathematical Functions

NumPy provides a comprehensive set of mathematical functions that operate element-wise on arrays. These functions are often much faster than implementing similar operations using Python loops, thanks to NumPy's optimized C implementations.

Let's look at some common examples:

```python
#Mathematical Function
```

```python
# Example array
arr_math = np.array([1, 4, 9, 16, 25])
print('Original Array:', arr_math)

# Square root
print('\nSquare Root (np.sqrt):', np.sqrt(arr_math))

# Exponential
print('Exponential (np.exp):', np.exp(arr_math))

# Logarithm (natural log)
print('Natural Logarithm (np.log):', np.log(arr_math))

# Sine (element-wise)
arr_angles = np.array([0, np.pi/2, np.pi])
print('\nArray of angles (radians):', arr_angles)
print('Sine (np.sin):', np.sin(arr_angles))

# Cosine (element-wise)
print('Cosine (np.cos):', np.cos(arr_angles))
```

```python
arr1 = np.array([[1, 2, 3], [4,5,6]])
arr2 = np.array([[7, 8, 9], [10,11,12]])

print('Array 1:\n', arr1)
print('Array 2:\n', arr2)

print('\nAddition:\n', arr1 + arr2)
print('\nSubtraction:\n', arr1 - arr2)
print('\nMultiplication:\n', arr1 * arr2)
print('\nDivision:\n', arr1 / arr2)
```

## NumPy Broadcasting

NumPy broadcasting is a powerful mechanism that allows NumPy to work with arrays of different shapes when performing arithmetic operations. It effectively 'stretches' the smaller array across the larger array so that they have compatible shapes without actually making copies of the data.

### How Broadcasting Works:

For two arrays to be compatible for broadcasting, their dimensions must satisfy one of the following conditions, starting from the trailing dimension (rightmost):

1.  **Equal:** The dimensions are equal.
2.  **One is 1:** One of the dimensions is 1.

If these conditions are not met, a `ValueError` is raised, indicating that the arrays are not compatible for broadcasting.

```python
#Make a brodcasting in depth alone with numpy and array of differcient by 1d and 2d
```

```python
print('--- Broadcasting with 1D Arrays (Addition) ---')

arr_1 = np.array([1, 2, 3,4])
arr_2 = np.array([5])
sum_arr = arr_1 + arr_2

print('Array 1:', arr_1)
print('Array 2:', arr_2)

print('\nResult of arr_1 + arr_2 (Broadcasting):', sum_arr)
```

```python
print('--------------Broadcasting with 2D array (Addition)----------------')


arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
arr_2d2 = np.array([[10, 20, 30]])

sum_arr = arr_2d + arr_2d2

print('Array 2D:\n', arr_2d)
print('Array 1D:', arr_2d2)
```

```python
print('------------------Broadcasting with 2d and 3d array(addition)------------')

arr_3D = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
arr_2D = np.array([[10, 20], [30, 40]])

sum_arr = arr_3D + arr_2D

print('Array 3D:\n', arr_3D)
print('Array 2D:\n', arr_2D)
print('sum_arr:\n', sum_arr)
```

```python
print('-------------Broadcasting 1d*scalar(multiplication)------------')
arr_1 = np.array([1,2,3,4])
scalar = (4)

result_mult = arr_1*scalar
print('Array 1:', arr_1)
print('Scalar:', scalar)
print('\nResult of arr_1 * scalar (Broadcasting):', result_mult)
```

```python
print('-------------Broadcasting 2d*scalar(multiplication)------------')
arr_2 = np.array([[1, 2, 3], [4, 5, 6]])
scalar =(6)

result_mult = arr_2*scalar
print('Array 2D:\n', arr_2)
print('Scalar:', scalar)
print('\nResult of arr_2 * scalar (Broadcasting):\n', result_mult)
```

```python
print('-------------Broadcasting  3d array*scalar(multiplication)------------')
arr_3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
scalar =(10)
result_mult = arr_3*scalar
print('Array 3D:\n', arr_3)
print('Scalar:', scalar)
print('\nResult of arr_3 * scalar (Broadcasting):\n', result_mult)
```

String function

```python
# Example 1: Convert to uppercase
string_array = np.array(['hello', 'world', 'numpy'])
uppercase_array = np.char.upper(string_array)
print('Original array:', string_array)
print('Uppercase array:', uppercase_array)

# Example 2: Join strings
join_array = np.array([['hello', 'world'], ['python', 'numpy']])
joined_string = np.char.join(' ', join_array) # Join elements within each sub-array
print('\nArray for joining:', join_array)
print('Joined strings:', joined_string)

# Example 3: Check for prefix
prefix_check = np.char.startswith(string_array, 'n')
print('\nCheck for prefix "n" in', string_array, ':', prefix_check)

# Example 4: Replace substring
replace_array = np.char.replace(string_array, 'o', '0')
print('\nOriginal array:', string_array)
print('Array after replacing "o" with "0":', replace_array)
```
